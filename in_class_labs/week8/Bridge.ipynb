
# Cross-Chain Bridges

## Introduction to Cross-Chain Interoperability
Blockchains generally operate in isolated environments, meaning an asset or smart contract on one chain cannot natively interact with another. Cross-chain bridges solve this by facilitating the transfer of information and digital assets between different blockchain networks. 



## Standard Bridge Mechanics (Committee Model)
When transferring assets across chains (e.g., from an Ethereum testnet to a Polygon testnet), tokens typically do not "move" across networks. Instead, a lock-and-mint mechanism is used:
1. A user deposits tokens into a specific smart contract on the source chain.
2. This contract notifies a committee of bridge managers (operating off-chain) that a transaction has occurred.
3. The committee votes to approve a corresponding transaction on the destination chain.
4. A wrapped version of the asset is minted or released to the user's wallet on the destination chain.

## Cross-Chain Bridge Vulnerabilities
Because bridges hold large amounts of liquidity and rely on complex off-chain/on-chain interactions, they are prime targets for malicious actors. To date, cross-chain bridges have been hacked for more than $2.8 billion—representing almost 40% of the entire value hacked in Web3, according to DefiLlama. 

Here are three typical bridge hacks:
- [Ronin Bridge (March 2022)](https://www.halborn.com/blog/post/explained-the-ronin-hack-march-2022): Five out of nine private keys needed to facilitate, approve, and execute transactions on the Ronin Bridge multisig were compromised. The Ronin Network adopted a 5-of-9 multisignature mechanism, requiring five valid validator signatures to authorize deposits or withdrawals. Attackers first compromised Sky Mavis' systems, gaining control of four validators owned by Sky Mavis. They then exploited an unrevoked allowlist to generate a signature from a third-party validator controlled by Axie DAO. With five valid signatures in hand, the attackers authorized two malicious withdrawals, stealing approximately $624 million.

- [Harmony Bridge (June 2022)](https://www.halborn.com/blog/post/explained-the-harmony-horizon-bridge-hack): Two out of the five private keys needed to facilitate, approve, and execute transactions on the Harmony Bridge multisig were compromised. The Harmony Horizon Bridge used a 2-of-5 multisignature scheme, meaning only two validator private keys were required to approve any transaction. Attackers successfully accessed and decrypted two of the bridge's private keys—despite the keys being encrypted with passphrases and a key management service. With these two keys, the attackers created and confirmed a malicious transaction, stealing $100 million in tokens, which were later laundered through Tornado Cash.
 
 - [Socket Interoperability Protocol (January 2024)](https://sockettech.notion.site/Socket-Incident-Report-16-Jan-9aba3bbf08814fc49e4f2ffb58284912): A flaw in the protocol smart contracts affected wallets with infinite approvals to Socket smart contracts. The vulnerability existed in the newly added WrappedTokenSwapperImpl module, which had an unverified contract call that failed to properly validate user inputs. Attackers exploited this flaw by combining a zero-amount swap with malicious swapExtraData, allowing them to call the transferFrom function of token contracts that had granted infinite approvals to the Socket Gateway. The attack was carried out via two malicious transactions on Ethereum, stealing approximately $3.3 million.

## Send Assets Across Different Blockchains
**Objective:** Transfer testnet ETH from the Sepolia chain (Ethereum) to the Amoy chain (Polygon) to understand standard bridge flow.

**Instructions:**
1. Go to the GUI for the [bridge between Sepolia and Amoy](https://portal.polygon.technology/bridge) and connect your MetaMask wallet.
2. Select the **ETH token** and the **Sepolia testnet**.
3. Enter the amount of ETH you wish to transfer (e.g., **0.01 ETH**).
4. Click **Transfer**.
5. Confirm the transaction via MetaMask.
6. **Verification:** View your transaction on Etherscan. If you go to the transactions tab on the Polygon portal, you should be able to see a pending transaction. Clicking on it will reveal the Sepolia transaction hash. After several minutes, you may view your Amoy balance on https://amoy.polygonscan.com/. 
7. **Observation:** Notice that the transaction transfers your ETH to a [certain smart contract](https://sepolia.etherscan.io/address/0x930c824c1e423a4b4949c665c4e92bd8f6ccf04e). This is the contract that locks your funds and alerts the off-chain bridge managers to approve the minting on the Amoy chain.
