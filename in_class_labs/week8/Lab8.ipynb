## The Vulnerable Lending Protocol

Below is the core logic of the lending protocol: (Note that, for simplicity, we have removed some parts of the code that are only for the purpose of the lab. Only the admin can use those parts, you don't need to worry about them. However, you can see the full code on Etherscan.)

```javascript
// SPDX-License-Identifier: MIT
pragma solidity ^0.8.17;

import "@uniswap/v3-core/contracts/interfaces/IUniswapV3Pool.sol";
import "@uniswap/swap-router-contracts/contracts/interfaces/IV3SwapRouter.sol";
import "@uniswap/lib/contracts/libraries/TransferHelper.sol";

interface IERC20 {
    function balanceOf(address) external view returns (uint256);
    function transfer(address, uint256) external returns (bool);
    function transferFrom(address, address, uint256) external returns (bool);
    function approve(address, uint256) external returns (bool);
}

interface IFlashLoanReceiver {
    function executeFlashLoan(uint256 amount) external;
}

contract LendingProtocol {
    IERC20 public immutable USTETH; // 0x4E22e9951770b98d4f3B2BA6647f0f40A15A4057
    IERC20 public immutable USTUSD; // 0xc83B0efA5B3F13851DfA11de72EF6AFeF026730c
    IUniswapV3Pool public immutable pricePool; // 0x8115b454E921534ec52f78A5473756cEF2dB0A30
    
    mapping(address => uint256) public collateralBalance;
    mapping(address => bool) public blocked;
    address public reserveAddress;
    uint256 public initialPoolSize; // 200000000000000000000000000
    uint256 public totalReserveAmount; // 5000000000000000000000000000
    uint256 public flashLoanReserve; // 100000000000000000000000000 (100 million USTUSD reserved for flash loans)
    uint256 public USTUSDAvailableLiquidity; // USTUSD liquidity available for borrowing
    address private constant SWAP_ROUTER = 0x3bFA4769FB09eefC5a80d6E87c3B9C650f7Ae48E;

    IV3SwapRouter public immutable swapRouter = IV3SwapRouter(SWAP_ROUTER);

    event Borrowed(address indexed user, uint256 amount);
    event ExploitDetected(address indexed exploiter, uint256 amount);
    event FlashLoanExecuted(address indexed borrower, uint256 amount);

    constructor(
        address _USTETH, 
        address _USTUSD, 
        address _pool, 
        uint256 _initialPoolSize, 
        uint256 _reserveAmount,
        uint256 _flashLoanReserve
    ) {
        USTETH = IERC20(_USTETH);
        USTUSD = IERC20(_USTUSD);
        pricePool = IUniswapV3Pool(_pool);
        
        reserveAddress = msg.sender;
        initialPoolSize = _initialPoolSize;
        totalReserveAmount = _reserveAmount;
        flashLoanReserve = _flashLoanReserve;
    }

    function initialFunding() external {...}

    function depositCollateral(uint256 collateralAmount) external {
        require(!blocked[msg.sender], "blocked");
        require(USTETH.transferFrom(msg.sender, address(this), collateralAmount), "Collateral transfer failed");
        collateralBalance[msg.sender] += collateralAmount;
    }

    function getUSTETHPrice() public view returns (uint256) {
        (uint160 sqrtPriceX96, , , , , , ) = pricePool.slot0();
        uint256 price = uint256(sqrtPriceX96) * uint256(sqrtPriceX96) >> (96 * 2);
    
        return price;
    }

    function borrow(uint256 amountUSTUSD) external {
        require(!blocked[msg.sender], "blocked");

        uint256 collateralValue = collateralBalance[msg.sender] * getUSTETHPrice();
        require(collateralValue >= amountUSTUSD, "Insufficient collateral value");
        require(USTUSDAvailableLiquidity >= amountUSTUSD, "Not enough liquidity");

        USTUSD.transfer(msg.sender, amountUSTUSD);
        USTUSDAvailableLiquidity -= amountUSTUSD;
        emit Borrowed(msg.sender, amountUSTUSD);

        // If exploited (full pool drain), fix the pool price and trigger auto-refill
        ...
    }

    function requestFlashLoan(uint256 amount) external {
        require(amount <= flashLoanReserve, "Insufficient flash loan reserve");

        // Send the funds to the borrower
        USTUSD.transfer(msg.sender, amount);
        IFlashLoanReceiver(msg.sender).executeFlashLoan(amount);

        // Ensure full repayment
        require(USTUSD.balanceOf(address(this)) >= flashLoanReserve, "Flash loan not fully repaid");

        emit FlashLoanExecuted(msg.sender, amount);
    }

    function refund() external {...}
}
```

### Protocol Introduction

#### Overview: 
- The lending protocol allows users to deposit `USTETH` as collateral and borrow `USTUSD` based on the collateral's value. The price of `USTETH` is determined by a Uniswap V3 pool. 
- It determines the collateral’s value using a **price oracle**, which is the current **exchange rate** from a `USTETH/USTUSD` decentralized exchange pair (the Uniswap V3 pool).
- There is no administrator setting the price; the contract simply trusts the DEX’s spot price. 

#### Key Functions:
- `depositCollateral(uint256 collateralAmount)`: Users can deposit `USTETH` as collateral.
- `borrow(uint256 amountUSTUSD)`: Users can borrow `USTUSD` based on the value of their collateral. The protocol checks the collateral's value using the price from the Uniswap V3 pool. If the collateral value is sufficient, it allows the borrow.
- `getUSTETHPrice()`: This function retrieves the current price of `USTETH` in terms of `USTUSD` from the Uniswap V3 pool. It uses the `sqrtPriceX96` value from the pool's `slot0()` function to calculate the price.

#### Contract behavior
When you call the lending contract’s borrow function, it roughly does the following:
- **Checks your collateral value:** It queries the USTETH/USTUSD pair for reserves to compute the price of USTETH. For example, if the pair has `(reserveUSTETH, reserveUSTUSD)`, the price of 1 USTETH (in USTUSD) might be calculated as `price = reserveUSTUSD / reserveUSTETH`. Then `collateralValue = yourUSTETHBalance * price`. (No sanity checks or time averaging are applied.)
- **Ensures sufficient collateral:** If your collateral value (in USTUSD) is *at least* the amount of USTUSD you want to borrow, it proceeds. (For simplicity, assume a 1:1 collateral ratio requirement – no extra margin – making the exploit even easier.)
- **Transfers USTUSD to you:** If the check passes, the contract transfers the requested USTUSD from its pool to your address, as a loan.


#### Why it's vulnerable
Using a live DEX price directly is dangerous because an attacker can **manipulate the price** with trades. The contract doesn’t use any time-weighted average or external oracle – it just reads the instantaneous pool ratio. This means a sudden buy or sell on the USTETH/USTUSD pool can wildly skew the reported price. An attacker can pump the USTETH price to absurd levels *within one transaction*, tricking the lending contract into thinking the collateral is worth far more USTUSD than it actually is. As one security analysis notes, protocols that naively use a liquidity pool’s spot price as their oracle are *"almost 100%" guaranteed to get rekt by a flash loan or price manipulation attack* ([The Full Guide to Price Oracle Manipulation Attacks](https://www.cyfrin.io/blog/price-oracle-manipulation-attacks-with-examples#:~:text=All%20these%20questions%20fit%20critically,with%20a%20larger%20market%20cap)).



#### Key vulnerability
The `getUSTETHPrice()` function simply uses the DEX reserve ratio as the price. If you can drastically lower `reserveUSTETH` relative to `reserveUSTUSD` (by buying up USTETH from the pool), the calculated price skyrockets. The contract will think your collateral is worth a fortune in USTUSD, allowing you to borrow the **entire USTUSD pool** with a minimal amount of USTETH deposited. There are no oracles like Chainlink or time-weighted average prices (TWAP) to prevent this – it’s entirely trusting the manipulable spot price ([The Full Guide to Price Oracle Manipulation Attacks](https://www.cyfrin.io/blog/price-oracle-manipulation-attacks-with-examples#:~:text=,downside%20is%20that%20they%20are)).

## Exploit Strategy Overview

To exploit this lending protocol, the high-level strategy is:

1. **Acquire USTETH cheaply, then inflate its price.** Use USTUSD to buy a large amount of USTETH from the USTETH/USTUSD pool. This trade will remove USTETH from the pool and put in USTUSD, causing the pool price of USTETH to shoot up (because now **each USTETH is backed by more USTUSD in the pool**). This swap is done by taking a flashloan of USTUSD, so, you don't need to own any USTUSD beforehand.

2. **Borrow USTUSD against overpriced USTETH collateral.** With the price now artificially high, even a small USTETH deposit is valued extremely high by the lending contract. You call the borrow function to take as much USTUSD as possible. The goal is to **drain the USTUSD pool** (the “jackpot”). Since the contract thinks your collateral is ample, it will give you all its USTUSD.

3. (Optional for profit realism) You could swap back the USTETH for USTUSD after borrowing, to return the price to normal and pocket the difference. However, in this simplified competition, you don’t actually need to do this to succeed – draining the pool is enough to win the round.

**Why a contract?** While you could attempt the steps manually (buy USTETH on a DEX, then call borrow), using a **custom exploit contract** has advantages:
- It lets you sequence actions within one transaction or script, making the exploit faster and more automated (important if racing other attackers).
- You can use flashloans (a flashloan needs to be borrowed and paid back in the same transaction) and perform the attack without owning a lot of tokens beforehand.
- It can hold tokens and interact with multiple contracts (DEX and lending) in one place.
- You can easily reuse the exploit code for each attempt by just deploying a new instance (if your first address gets blocked).

## Jackpot Rounds and blocking Mechanism

This lending contract is set up to allow **multiple exploits in succession**, so that after one student succeeds, others still have a chance (with diminishing rewards). Here’s how it works:

- **Initial round:** The instructor funded the contract with the full original USTUSD amount (the initial jackpot). The first student to exploit the contract will withdraw most or all of these funds. The contract then **blocks the exploiter’s address** (so the same student cannot simply repeat).

- **USTETH price back to normal:** After each drain, the contract brings the price of the Uniswap pool back to normal. It does so by swapping the abandoned USTETH (that was deposited as collateral in the lending contract) for USTUSD.

- **Refill:** The available USTUSD liquidity to borrow gets refilled in the lending contract. Now a second student can perform the same exploit. The second exploiter’s address will then be blocked as well.

- **Subsequent rounds:** The pattern continues till all the students have exploited the contract. 

**blocking:** The contract uses a mapping to permanently record addresses that have already won. If a blocked address tries to interact (deposit or borrow), the contract will reject them.

*⚠️ Note:* All the exploits are happening on a public test network. Be mindful that anyone with the contract address and know-how could attempt the exploit. In a classroom setting, this is fine (and part of the fun), but remember on real networks this kind of vulnerability is catastrophic. Real attackers bundle steps into a single atomic transaction (often with flash loans) to ensure they cannot be outpaced or the price returns to normal before they act.

## Challenge: Defending Against Oracle Manipulation

- **Oracle Design:** What if the protocol used a more reliable price feed instead of a raw DEX price? - **Price Averaging:** DEXes like Uniswap v2/v3 offer a **Time-Weighted Average Price (TWAP)** oracle. This means the price is averaged over a window of time, making rapid swings (like those from flash trades) much less effective. Flash loan attacks usually manipulate price within one block, but a TWAP would **smooth out the price**, making the oracle **resistant to instantaneous manipulation* ([The Full Guide to Price Oracle Manipulation Attacks](https://www.cyfrin.io/blog/price-oracle-manipulation-attacks-with-examples#:~:text=,lagging%20indicator%20of%20price%20values))】. - **Collateral and Liquidity Safeguards:** How might the lending protocol itself limit damage? Think about adding **collateralization ratios** (requiring more collateral than the value of the loan), **borrow limits**, or **circuit breakers** that pause borrowing if prices move too fast. Could requiring multiple price sources or a delay before using a new price update help?
