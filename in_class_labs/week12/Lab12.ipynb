# Register an ENS

Ethereum Name Service is a decentralized domain name service that runs on the Ethereum blockchain; it maps a human-readable name to an Ethereum address (EOA or a contract). It improves usability since ENS is well integrated within wallets - making it easier to transfer funds to your friends or interact with known contracts. In this part of the lab, you will set up an ENS name for one of your wallets on Sepolia. Please follow the steps below:

- Go to the ENS web app at https://sepolia.app.ens.domains/
- Type in a preferred name and pick it if available 
  - Note that the name will be available in public, so choose a pseudonymous name so that your wallet address cannot be linked back to your real name. For example: `FTEC5320t1.eth`
  - ‼️Remember the name you choose, since it will be used in the following steps.
- Start the timer: the timer ensures that there is no front-running attack on ENS. 
  - Wait for at least 1 minute before you proceed to the next step. You can check the timer on the ENS web app.
- Register the name after the wait. 
  - Example transaction: https://sepolia.etherscan.io/tx/0x12cefa30dbe029c09edab69f9c40ef86f8d1e5953dcb49259cd1d6de3db3c73a 
  - You will find that it mint a new `ENS NFT(ERC-721)` to your wallet, which represents the ownership of the ENS name. 
- Try to send some SepoliaETH to your ENS name on Metamask (e.g., `FTEC5320t1.eth`). 

How does the timer mechanism prevent front running? Hint: The address who submits the commit and the reveal first gets the name, and there should be a minimum 1-minute gap between the commit and reveal

# Interacting with element.market on Sepolia testnet

- https://testnets.element.market/ethereum 

## Create and deploy your own collection

1. Connect wallet and go to profile: `My NFTs`
2. Check out your NFTs. 
   - Uniswap V3 is the NFTs representing V3 LP positions you took in labs earlier.
   - You can also find the ENS NFT you just minted in the previous part of the lab.
3. Create a `drop collection` and mint some NFTs. 
   - Click `Create` on the top Navbar. 
   - Fill in the collection name, description, symbol, and image. 
   - You can set the `Mint Start Time` to be current time.
   - You can set the Mint Price (e.g., 0.01 ETH) under the `Mint Stage` and click `Done` to save the mint stage.
4. Finally, click the button to deploy the collection. The website will ask you to sign a few transactions to deploy the collection.
   - You can publish the collection after deployment. Then, other users can find your collection and mint from it.

## Mint others' collection
- Go to the collection we created for this class: https://testnets.element.market/drop/ftec5320-test-1 
- Mint an NFT from the collection. 
- View your NFT in your profile. (You can also view the NFT on Etherscan which will be faster than Element's profile page to update the ownership change.)

## List and buy NFTs
- Go to your profile `My NFTs` and list one of your NFTs for sale.
  - You can modify the price and set an expiration time for the listing.
- Click the `DETAILS` of the listing and share the link with your friend. Or you can also use another account to buy the NFT from your listing.
- After the purchase, check the ownership change of the NFT on Etherscan or Element's profile page.

<img src="./NFT Figs/fig2.png" alt="list NFT" height="400"/>

## Send an NFT by resolving the ENS name
- Go to your profile `My NFTs` and click `Send` on one of your NFTs.
  - If you already saled the NFT in the previous step, you can also go to the [collection page](https://testnets.element.market/drop/ftec5320-test-1) and `Mint` a new one to send.
- Go to ENS (https://sepolia.app.ens.domains/) and resolve your friend's or your another account's ENS name to get the address.
- Send the NFT to the address you get from ENS resolution.
- Check the ownership change of the NFT on Etherscan or Element's profile page.
