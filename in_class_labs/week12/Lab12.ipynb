# Register an ENS

Ethereum Name Service is a decentralized domain name service that runs on the Ethereum blockchain; it maps a human-readable name to an Ethereum address (EOA or a contract). It improves usability since ENS is well integrated within wallets - making it easier to transfer funds to your friends or interact with known contracts. In this part of the lab, you will set up an ENS name for one of your wallets on Sepolia. Please follow the steps below:

- Go to the ENS web app at https://sepolia.app.ens.domains/
- Type in a preferred name and pick it if available 
  - Note that the name will be available in public, so choose a pseudonymous name so that your wallet address cannot be linked back to your real name. For example: `FTEC5320t1.eth`
  - ‼️Remember the name you choose, since it will be used in the following steps.
- Start the timer: the timer ensures that there is no front-running attack on ENS. 
  - Wait for at least 1 minute before you proceed to the next step. You can check the timer on the ENS web app.
- Register the name after the wait. 
  - Example transaction: https://sepolia.etherscan.io/tx/0x12cefa30dbe029c09edab69f9c40ef86f8d1e5953dcb49259cd1d6de3db3c73a 
  - You will find that it mint a new `ENS NFT(ERC-721)` to your wallet, which represents the ownership of the ENS name. 
- Try to send some SepoliaETH to your ENS name on Metamask (e.g., `FTEC5320t1.eth`). 

How does the timer mechanism prevent front running? Hint: The address who submits the commit and the reveal first gets the name, and there should be a minimum 1-minute gap between the commit and reveal

Please follow the steps below:

- Visit this contract, specifically looking at `Doodles.sol` and `ERC721.sol`.
- Specifically, note and understand the connection between the `setBaseURI` method in `Doodles.sol` and the `tokenURI` method in `ERC721.sol`, which is what allows the storage of a URI for each token's metadata.
- Access the `read contract` tab, and call the `tokenURI` method on a token id of your choice (e.g., `1`).
- Example output for token id `1`: `ipfs://QmPMc4tcBsMqLRuCQtPmPe84bpSjrC3Ky7t3JWuHXYB4aS/1`

  - Example:
  https://ipfs.io/ipfs/QmPMc4tcBsMqLRuCQtPmPe84bpSjrC3Ky7t3JWuHXYB4aS/1

**IPFS (InterPlanetary File System)** is a decentralized storage network that allows for the storage and sharing of data in a distributed manner. It is often used in conjunction with NFTs to store the metadata and media associated with the NFT, as it provides a more scalable and cost-effective solution compared to storing data directly on the blockchain.

# Minting an on-chain NFT from contract class provided

We have created a custom NFT contract for this class; it generates SVG images using on-chain logic associated with the token ID to be minted.

- Access our ERC-721 NFT contract at: https://sepolia.etherscan.io/address/0x16852eB1F7b747318b769FaF87bcF65fFa721F47#writeContract 
- Using `mintNFT` to mint your own NFT. This method requires `0.01` ether. 
  - You will get the `tokenID` of that NFT after minting. You can view your tokenID in the transaction receipt on Etherscan. 
- Go to [Read Tab](https://sepolia.etherscan.io/address/0x16852eB1F7b747318b769FaF87bcF65fFa721F47#readContract) and put your tokenID in the `tokenURI` method to access the on-chain metadata and generate the image. It should show a long string that starts with `data:image/svg+xml;base64,`, which is a base64 encoded string that contains the SVG image data. 
- Copy the output string generated from the `tokenURI` method and decode it using the following python decoder.


````python
import base64
import json

# Your Base64 JSON string goes here
tokenURI_base64 = "" # to be filled with the output of tokenURI method on Etherscan

# Remove the data URL prefix if present
if "," in tokenURI_base64:
    tokenURI_base64 = tokenURI_base64.split(',', 1)[1]

# Decode the JSON string
decoded_json_bytes = base64.b64decode(tokenURI_base64)
decoded_json = json.loads(decoded_json_bytes)

# Extract the Base64 string of the image from the decoded JSON
image_base64 = decoded_json['image']

# Remove the Base64 prefix from the image data
if "," in image_base64:
    image_base64 = image_base64.split(',', 1)[1]

# Decode the image data
image_data = base64.b64decode(image_base64)

# Save to a file
image_filename = "NFT_metadata.svg"  # Saved as SVG since MIME type is image/svg+xml
with open(image_filename, "wb") as file:
    file.write(image_data)

print(f"Image has been saved as {image_filename}")
````

# Introduction to IPFS

Off-chain storage is critical to the NFT ecosystem. NFTs being unique, prefer IPFS since it is a content-addressable storage platform. We will look at how to interact with IPFS using Pinata.  

- Access Pinata IPFS platform by going to [https://app.pinata.cloud/pinmanager](https://app.pinata.cloud/pinmanager) and signing up for an account.

- Note the `CID` generated by the image.
